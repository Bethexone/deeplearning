```python
import torch
from torch import nn
from torch.nn import functional as F
torch.device('cuda'),torch.cuda.device('cuda'),torch.cuda.device_count()
```

```python
def try_gpu(i=0):
    if torch.cuda.device_count() >= i+1:
        return torch.device(f'cuda:{i}')
    return torch.device('cpu')
```

```python
def try_all_gpus():
    """返回所有可用的GPU，若没有GPU，则返回cpu"""
    devices = [torch.device(f'cuda:{i}')for i in range(torch.cuda.device_count())]
    return devices if devices else [torch.device('cpu')]
```

```python
try_all_gpus()
```

```python
x = torch.tensor([1.,2.,3.,4.])
x.device
```

```python
x = torch.tensor([1.,2.,3.,4.],device =try_gpu() )
x.device
```

```python
y = torch.tensor([1.,2.,3.,4.],device =try_gpu())
y.device
```
