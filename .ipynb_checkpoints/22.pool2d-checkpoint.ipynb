### 池化层

```python
import torch
import torch.nn as nn
```

```python
pool2d = nn.MaxPool2d(3)
```

```python
X = torch.arange(16,dtype = torch.float32).reshape((1,1,4,4))
```

```python
pool2d(X)
```

```python
pool2d = nn.MaxPool2d((2,3),padding=1,stride=2)
```

```python
pool2d(X)
```

```python
pool2d = nn.MaxPool2d((2,3),padding=(1,1),stride=(2,3))
```

```python
pool2d(X)
```
