```python
import torch
import time
from torch import nn
from torch.nn import functional as F
from py09_FashionMNIST_Dataset import get_fashion_mnist_labels,load_data_fashion_mnist
from deeplearn_tools import Animator,Accumulator,evaluate_accuracy,Timer
```

```python
class Reshape(nn.Module):
    def forward(self,X):
        return X.view(-1,1,28,28)
        
```

```python
class LeNet(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.net = nn.Sequential(
            
            nn.Conv2d(1,6,kernel_size=5,padding=2),nn.Sigmoid(),
            nn.AvgPool2d(kernel_size=2,stride=2),
            nn.Conv2d(6,16,kernel_size=5),nn.Sigmoid(),
            nn.AvgPool2d(kernel_size=2,stride=2),nn.Flatten(),
            nn.Linear(16*5*5,120),nn.Sigmoid(),
            nn.Linear(120,84),nn.Sigmoid(),
            nn.Linear(84,10)
        )

    def forward(self,X):
        return self.net(X)
        

    def Print_layerinfo(self,X):
        for layer in self.net:
            X = layer(X)
            print(layer.__class__.__name__,'output shape:\t',X.shape)
```

```python
X = torch.rand(size=(1,1,28,28),dtype=torch.float32)
X.shape
```

```python
letnet_module = LeNet()
letnet_module.Print_layerinfo(X)
```

```python
num_epoches = 10
batch_size=256
train_iter,test_iter = load_data_fashion_mnist(batch_size=batch_size)
```

```python
optimizer = torch.optim.SGD(letnet_module.parameters(),lr=0.1)
loss = nn.CrossEntropyLoss()
```

```python
def train(train_iter,test_iter,net,Optimizer,Loss,num_epoches,device='cuda'):
    def init_weights(layer):
        if type(layer)== nn.Linear or type(layer) == nn.Conv2d:
            nn.init.xavier_uniform_(layer.weight)
    net.apply(init_weights)
    net.to(device)
    print('training on',device)
    ani = Animator(xlabel='num_epoch',xlim=[1,num_epoches],legend=['train_loss','train_accuracy','test_accuracy'],yscale='log')
    num_batches = len(train_iter)
    for epoch in range(num_epoches):
        start_time = time.time()
        accum = Accumulator(3)
        l_sum = 0
        for X,y in train_iter:
            X,y = X.to(device),y.to(device) 
            y_hat = net(X)
            l = Loss(y_hat,y)
            Optimizer.zero_grad()
            l.backward()
            Optimizer.step()
            l_sum += l
        end_time = time.time()
        print(epoch+1,': ',round(end_time-start_time,3))
        train_acc = evaluate_accuracy(net,train_iter,device)
        test_acc = evaluate_accuracy(net,test_iter,device)
        ani.add(epoch+1,[l_sum,train_acc,test_acc])
        print(f"{epoch}:train loss,{l_sum},train_acc:{train_acc},test_acc:{test_acc},time:{epoch_time}")
```

```python
train(train_iter,test_iter,letnet_module,optimizer,loss,num_epoches)
```
