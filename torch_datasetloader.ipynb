### **Dataset** 是一个抽象类，需要重写其 __\_\_getitem\_\_(self, index)__ 和 __\_\_len\_\_(self)__

```python
class MyDataset(Dataset):
    def __init__(self):
        super().__init__()
        self.x = np.random.random((10,10))
        self.w = np.random.random(10)
        self.y = self.x *self.w
    def __len__(self):
        return len(self.x)
    def __getitem__(self,idx):
        return self.x[idx,:],self.y[idx,:]
        print(idx)
    
```

```python
data = MyDataset()
```
